```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, cross_validate
```

```python
df = pd.read_csv('Travel_usar.csv')
```

```python
data = df.copy()
data['Gender'] = data['Gender'].replace('Fe Male', 'Female')
data = data.drop(columns=['CustomerID','Designation'])
```

```python
# Definir target y variables, separar train y test estratificado
X = data.drop(columns='ProdTaken')
y = data['ProdTaken']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
print(X_train.shape, X_test.shape)
```

```python
cols_out = ['Age', 'DurationOfPitch', 'NumberOfTrips', 'MonthlyIncome']

limites = {}

for c in cols_out:
    # Calculate Z-scores
    z_scores = (X_train[c] - X_train[c].mean()) / X_train[c].std()

    # Identify outliers (Z > 3 or Z < -3)
    outliers = z_scores.abs() > 3

    # Store lower and upper limits
    limites[c] = (
        X_train[c].mean() - 3 * X_train[c].std(),
        X_train[c].mean() + 3 * X_train[c].std()
    )

    # Count outliers
    print(c, 'outliers:', outliers.sum())
```
